```python
#Products.py
class product_category:
    def __init__(self) -> None:
        pass

    def get_category(self, id):
        return f"Category - {id}"
```

```python
#main.py

p = product_category()
res = p.get_category(id=2.5)
print(res)
```

```python
res1 = p.get_category(id="Electronics")
print(res1)
```
