`np.convolve` is a NumPy function used to perform **convolution** between two 1D arrays.

In signal processing and deep learning, convolution means:

> Slide one array over another, multiply matching values, and add them.

Syntax:

```python id="qvv6v4"
np.convolve(a, v, mode='full')
```

* `a` → first array
* `v` → second array (kernel/filter)
* `mode` → controls output size

---

# Simple Example

```python id="7xp1it"
import numpy as np

a = [1, 2, 3]
v = [0, 1, 0.5]

np.convolve(a, v)
```

Output:

```python id="6qes1o"
array([0. , 1. , 2.5, 4. , 1.5])
```

---

# How Convolution Works

We slide `v` across `a`.

## Step 1

```python id="pj5wqw"
[1, 2, 3]
[0]
```

Multiply and sum:

```python id="k9m5jk"
1*0 = 0
```

---

## Step 2

```python id="43p57d"
[1, 2, 3]
[0, 1]
```

Compute:

```python id="l4m32m"
1*1 + 2*0 = 1
```

---

## Step 3

```python id="hrw1e8"
[1, 2, 3]
[0, 1, 0.5]
```

Compute:

```python id="u8r6z5"
1*0.5 + 2*1 + 3*0
= 0.5 + 2
= 2.5
```

…and so on.

---

# Modes in `np.convolve`

## 1. `mode='full'` (default)

Returns complete convolution.

```python id="x8txjr"
np.convolve(a, v, mode='full')
```

Output length:

```python id="m9r6q2"
len(a) + len(v) - 1
```

---

## 2. `mode='same'`

Output has same size as first array.

```python id="2b10hx"
np.convolve(a, v, mode='same')
```

Useful in ML and image processing.

---

## 3. `mode='valid'`

Only returns positions where arrays fully overlap.

```python id="5lupum"
np.convolve(a, v, mode='valid')
```

Shortest output.

---

# Visualization

Signal:

```python id="ez2n7s"
a = [1, 2, 3]
```

# Common Uses

## Signal Processing

* smoothing signals
* filters
* audio processing

## Deep Learning / CNNs

* edge detection
* blur filters
* feature extraction

## Mathematics

* polynomial multiplication

Example:

```python id="8x2j0m"
np.convolve([1,2], [1,3])
```

Represents:

```python id="w4i4a7"
(1 + 2x)(1 + 3x)
```

Result:

```python id="qujff7"
[1, 5, 6]
```

which means:

```python id="rkh6kz"
1 + 5x + 6x²
```
