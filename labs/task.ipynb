## 1️⃣ Dataset Setup

Imagine you have a CSV file called `sales.csv` with columns:

```
OrderID, Date, CustomerID, Product, Category, Quantity, Price, Region
```

### Load the dataset:
```python
import pandas as pd

# Load CSV file
df = pd.read_csv("sales.csv")

# Preview first 5 rows
print(df.head())
```

---

## 2️⃣ Basic Operations

### Explore the dataset:
```python
# Check structure
print(df.info())

# Summary statistics
print(df.describe())

# Check for missing values
print(df.isnull().sum())

# Drop duplicates if any
df = df.drop_duplicates()
```

### Clean the dataset:
```python
# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Fill missing values in Quantity with 0
df['Quantity'] = df['Quantity'].fillna(0)

# Ensure Price is numeric
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
```

---

## 3️⃣ Intermediate Analysis

### Filtering:
```python
# Sales from 'North' region
north_sales = df[df['Region'] == 'North']
```

### Grouping & Aggregation:
```python
# Total revenue per product
df['Revenue'] = df['Quantity'] * df['Price']
product_sales = df.groupby('Product')['Revenue'].sum().reset_index()
print(product_sales)
```

### Pivot Tables:
```python
# Monthly sales by category
monthly_sales = df.pivot_table(
    values='Revenue',
    index=df['Date'].dt.to_period('M'),
    columns='Category',
    aggfunc='sum'
)
print(monthly_sales)
```

Your DataFrame has **51,290 rows** and **21 columns**. Most columns are already in good shape, but there are a few important datatype issues to fix before analysis.

### Key observations

* `sales` is stored as `str` → should likely be numeric (`float`)
* `order_date` and `ship_date` are stored as `str` → should be converted to datetime
* No missing values shown in any column
* Dataset looks like a retail/order management dataset similar to Superstore Global Sales

### Recommended datatype conversions

```python
import pandas as pd

# Convert dates
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])

# Convert sales to numeric
df['sales'] = pd.to_numeric(df['sales'], errors='coerce')
```

### Optional memory optimization

You can reduce memory usage significantly using categorical types:

```python
cat_cols = [
    'ship_mode', 'segment', 'state', 'country',
    'market', 'region', 'category',
    'sub_category', 'order_priority'
]

for col in cat_cols:
    df[col] = df[col].astype('category')
```
