```python
# Import liabraries

import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# 1. Define the model
class Sentiment(nn.Module):
    def __init__(self, vocab_size, emb_dim=64):
        super().__init__()
        self.emb = nn.Embedding(vocab_size, emb_dim)   # word embeddings
        self.fc  = nn.Linear(emb_dim, 1)               # classifier

    def forward(self, ids):
        x = self.emb(ids)              # (batch, tokens, emb_dim)
        x = x.mean(dim=1)              # average over tokens
        return self.fc(x).squeeze(-1)  # output logit
```

```python
# 2. Create dummy data (pretend reviews)
# Each review is represented as a list of token IDs
reviews = torch.tensor([
    [1, 5, 7, 9],   # "Great movie, loved it"
    [2, 6, 8, 1],      # "Terrible film, boring"
])
labels = torch.tensor([1., 0.])  # 1 = positive, 0 = negative
```

```python
# 3. Initialize model, loss, optimizer
model = Sentiment(vocab_size=10, emb_dim=64)
loss_fn = nn.BCEWithLogitsLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

```python
# 4. Training loop (tiny demo)
for epoch in range(10):
    logits = model(reviews)
    loss = loss_fn(logits, labels)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    print(f"Epoch {epoch}, Loss = {loss.item():.4f}")
```

```python
# 5. Prediction
with torch.no_grad():
    test_review = torch.tensor([[1, 2]])  # "Great, loved"
    score = model(test_review).item()
    print(score)
    sentiment = "Positive" if score > 0.5 else "Negative"
    print("Prediction:", sentiment)
```
